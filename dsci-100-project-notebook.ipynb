#### Task 3.1: Tidy the data (5 points)

```r
# The data itself is already tidy
# Select the variables needed for the question
womens_march_madness_tidy <- womens_march_madness |>
    select(conf_wins_pct, seed, tourney_finish) |>

# Remove rows with missing values
filter(!is.na(conf_wins_pct), !is.na(seed), !is.na(tourney_finish)) |>

# Change tournament finish into a categorical variable
mutate(tourney_finish = as_factor(tourney_finish))

# Show the first 6 rows of the tidy data
head(womens_march_madness_tidy)
```

```r
# Plot the distribution of conference win percentages
womens_march_madness_tidy |>
  ggplot(aes(x = conf_wins_pct)) +
  geom_histogram(bins = 20) +
  labs(title = "Distribution of Conference Win Percentages", x = "Conference Win Percentage (%)", y = "Number of Teams")
```

```r
# Calculate the proportion of each tournament finish
finish_proportions <- womens_march_madness_tidy |>
  group_by(tourney_finish) |>
  summarize(count = n()) |>
  mutate(proportion = count / sum(count))

# Plot the proportions
ggplot(finish_proportions, aes(x = proportion, y = tourney_finish)) +
  geom_bar(stat = "identity") +
  labs(title = "Proportion of Teams by Tournament Finish", x = "Proportion of Teams", y = "Tournament Finish")
```

Visualization 1: This hortizontal bar graph compares the average conference win percentage of teams based on how far they advanced in the tournament. It helps determine whether teams with higher conference win percentages advance further.

```r
# Calculate the average win percentage for each tournament finish
wins_summary <- womens_march_madness_tidy |>
  group_by(tourney_finish) |>
  summarize(avg_wins = mean(conf_wins_pct, na.rm = TRUE))

# Plot the average win percentage
wins_summary |>
  ggplot(aes(x = avg_wins, y = tourney_finish)) +
  geom_bar(stat = "identity") +
  labs(title = "Average Conference Win Percentage by Tournament Finish", x = "Average Conference Win Percentage", y = "Tournament Finish")
```

Visualization 2: This horizontal bar graph compares the average tournament seed of teams for different tournament finishes. Since lower seed numbers represent higher-ranked teams, it helps us determine whether stronger-seeded teams advance further.

```r
# Calculate the average seed for each tournament finish
seed_summary <- womens_march_madness_tidy |>
  group_by(tourney_finish) |>
  summarize(avg_seed = mean(seed, na.rm = TRUE))

# Plot the average tournament seed
seed_summary |>
  ggplot(aes(x = avg_seed, y = tourney_finish)) +
  geom_bar(stat = "identity") +
  labs(title = "Average Tournament Seed by Tournament Finish", x = "Average Tournament Seed", y = "Tournament Finish")
```
